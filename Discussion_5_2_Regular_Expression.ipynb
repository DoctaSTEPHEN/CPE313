```python
import pandas as pd

twitter = pd.read_csv("/content/drive/MyDrive/CPE 313/twitter_training.csv", header=None)
twitter.columns = ["Tweet_ID", "Entity", "Sentiment", "Tweet_content"]
twitter.head()
```

```python
import re

twitter['word_count'] = twitter['Tweet_content'].apply(lambda text: len(re.findall(r"on", str(text))))
twitter['char_count'] = twitter['Tweet_content'].apply(lambda text: len(re.findall(r"\,", str(text))))
twitter['num_count'] = twitter['Tweet_content'].apply(lambda text: len(re.findall(r"\w*3\b", str(text))))
```

```python
twitter.head()
```

```python
print(f'Total occurrences of "on": {twitter["word_count"].sum()}')
print(f'Total commas (,): {twitter["char_count"].sum()}')
print(f'Total occurrences of number "3": {twitter["num_count"].sum()}')
```

```python
URL_find = twitter['Tweet_content'].apply(lambda text: re.findall(r"\S+/\S+", str(text)))
URL = URL_find[URL_find.astype(bool)]
URL.head()
```
